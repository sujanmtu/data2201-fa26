## 2. Hypothesis testing fundamentals.

### What problem does hypothesis testing solve?

Suppose we've performed an experiment, or identified something interesting in our data.

- Say we've created a new vaccine.

- To assess its efficiency, we give one group the vaccine, and another a [placebo](https://en.wikipedia.org/wiki/Placebo). 

- We notice that the flu rate among those who received the vaccine is lower than among those who received the placebo (i.e. didn't receive the vaccine).

- One possibility: the vaccine doesn't actually do anything, and by chance, those with the vaccine happened to have a lower flu rate.

- Another possibility: receiving the vaccine made a difference – the flu rate among those who received the vaccine is lower than we'd expect due to random chance.

- **Hypothesis testing allows us to determine whether an observation is "significant."**

## What Is Hypothesis Testing?

**Hypothesis testing** is a statistical method used to evaluate a claim about a population using sample data.

- We begin with a claim about the population.
- We collect sample data and examine the evidence.
- We determine whether the observed results are consistent with the claim.

**Example:** A coin is claimed to be fair. If we toss it 100 times and observe 65 heads, is the coin fair, or is there evidence that it may be biased?

## Null and Alternative Hypotheses

Hypothesis testing begins with two competing statements:

- **Null Hypothesis ($H_0$):** The initial claim or assumption being tested.
- **Alternative Hypothesis ($H_1$):** The competing claim that we seek evidence to support.

**Example: Testing a Coin**

- $H_0$: The coin is fair ($P(\text{Heads}) = 0.5$).
- $H_1$: The coin is biased ($P(\text{Heads}) \neq 0.5$).

**Goal:** Determine whether the observed data provide sufficient evidence to reject $H_0$.

## How Does Hypothesis Testing Work?

A hypothesis test follows these basic steps:

1. **Define hypotheses:** State $H_0$ and $H_1$.
2. **Choose a test statistic:** Select a numerical measure to evaluate the observed data.
3. **Generate a null distribution:** Determine what results are expected if $H_0$ is true.
4. **Calculate the p-value:** Measure how unusual the observed result is under $H_0$.
5. **Make a decision:** Compare the p-value with a significance level ($\alpha$), commonly 0.05.

**Decision Rule:**

- If $p \leq 0.05$: Reject $H_0$.
- If $p > 0.05$: Fail to reject $H_0$.

**Remember:** Failing to reject $H_0$ does not prove that it is true.

### Test Statistic, p-value, and Statistical Decision

**Test Statistic:** A numerical measure used to evaluate the observed data under the null hypothesis ($H_0$).

**p-value:** The probability of obtaining a result at least as extreme as the observed result, assuming $H_0$ is true.

$$
\text{p-value} \approx
\frac{\text{Simulations at least as extreme as observed}}
{\text{Total number of simulations}}
$$

**Statistical Decision ($\alpha = 0.05$):**

Choose a **significance level** $\alpha$ before the test; we will use $\alpha=0.05$.

- **p-value $\leq 0.05$:** Reject $H_0$; sufficient evidence against the null hypothesis.
- **p-value $> 0.05$:** Fail to reject $H_0$; insufficient evidence against the null hypothesis.

**Example:** $p = 0.03$ → Reject $H_0$; $p = 0.08$ → Fail to reject $H_0$.

**Remember:** Failing to reject $H_0$ does not prove it is true.

## 3. Hypothesis Testing vs. Permutation Testing

**Hypothesis Testing (One Sample)**

We know the population distribution and have one sample.

> Is the observed sample consistent with the population distribution?

**Permutation Testing (Two Samples)**

We have two samples but do not know their population distributions.

> Do the two samples differ significantly, or could the observed difference be due to random chance?

**Key idea:** A permutation test is a type of hypothesis test that uses random shuffling to determine whether an observed difference between groups is statistically significant.

## 4. Coin-toss demo 
### Step 1 — Record the observation and set up the test
We observed **59 heads out of 100 tosses**. Under $H_0$, the chance of heads on each toss is 0.5; under $H_1$, it is greater than 0.5.

Run the setup cell. It records the observed statistic, significance level, and random-number generator. **Prediction:** Would you expect 59 heads to be common or uncommon if the coin is fair?

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n_tosses = 100
observed_heads = 59
alpha = 0.05
```

### Step 2 — Simulate one fair-coin experiment
Run the next cell to simulate **one set of 100 fair tosses** and count heads.

The result will generally differ from 50 and from the observed 59. **Ask students:** Why can a fair coin produce more than 50 heads?

```python
one_experiment = rng.binomial(n=n_tosses, p=0.5)
print(f"Heads in one simulated set of 100 tosses: {one_experiment}")
```

### Step 3 — Build the null distribution
One simulated experiment is not enough. Repeat the **100-toss experiment 10,000 times** under $H_0$ and record each head count.

Inspect the first 10 results. Each value represents a **different experiment of 100 tosses**, not an individual toss.

```python
simulated_heads = rng.binomial(n=n_tosses, p=0.5, size=10_000)
print("First 10 simulated results:", simulated_heads[:10])
```

### Step 4 — Visualize typical and unusual outcomes
Run the histogram. Its horizontal axis shows the **number of heads per 100 tosses**; its vertical axis shows how frequently each count occurred across simulated experiments.

Locate the observed **59 heads** relative to the center near 50. **Discuss:** Does the observed count fall in a common region or toward the high end of the null distribution?

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
bins = np.arange(simulated_heads.min() - 0.5, simulated_heads.max() + 1.5)
ax.hist(simulated_heads, bins=bins, density=True, color="steelblue", edgecolor="white", label="Fair-coin simulations")
ax.axvline(observed_heads, color="crimson", linestyle="--", linewidth=2, label="Observed: 59 heads")
ax.set(xlabel="Number of heads in 100 tosses", ylabel="Relative frequency", title="Null distribution: 10,000 fair-coin experiments")
ax.legend()
plt.tight_layout()
plt.show()
```

### Step 5 — Estimate the one-sided p-value
The expression `simulated_heads >= observed_heads` marks experiments with **59 or more heads**. Taking the mean of these Boolean values gives the fraction of simulations at least as extreme as ours.

**Interpretation:** If the coin were fair, approximately what percentage of 100-toss experiments would produce 59 or more heads? This is *not* the probability that the coin is fair.

```python
p_value = np.mean(simulated_heads >= observed_heads)
print(f"Simulated p-value: {p_value:.4f} ({p_value:.2%})")
```

### Step 6 — Make the decision and communicate it
The original code compares the estimated p-value with **$\alpha=0.05$**. Read the printed result, then state the conclusion in context.

- **Reject $H_0$:** Evidence supports the claim that the coin favors heads.
- **Fail to reject $H_0$:** The observed result does not provide sufficient evidence that the coin favors heads; it does not prove fairness.

**Check:** Why would changing the observed count from 59 to 65 affect the p-value?

```python
if p_value < alpha:
    print("Reject H0: the results provide evidence that the coin favors heads.")
else:
    print("Fail to reject H0: insufficient evidence that the coin favors heads.")
```

## Wrap-up: the complete reasoning chain
**Question → $H_0$ and $H_1$ → statistic → null distribution → p-value → decision → conclusion.**
