### 1. Sampling in practice
Use `np.random.choice` to sample from a sequence, `.sample()` for DataFrames, and `np.random.multinomial` for categorical counts.

## 2. Hypothesis Testing Demo: Total variation distance.

### Ethnic distribution of California vs. UCSD
Compare the ethnic distributions of California residents and UCSD students. Is the difference larger than we would expect from random sampling?

```python
eth = pd.DataFrame(
    [['Asian', 0.15, 0.51],
     ['Black', 0.05, 0.02],
     ['Latino', 0.39, 0.16],
     ['White', 0.35, 0.2],
     ['Other', 0.06, 0.11]],
    columns=['Ethnicity', 'California', 'UCSD']
).set_index('Ethnicity')

eth
```

### Hypotheses
- **$H_0$:** UCSD students are sampled from the California distribution.
- **$H_1$:** UCSD students are not sampled from that distribution.
We use TVD to measure the difference.

```python
eth.plot(kind='barh', title='Ethnic Distribution of California and UCSD', barmode='group')
```

### Total variation distance (TVD)
TVD measures the difference between two categorical distributions: $\frac12\sum |p_i-q_i|$. A larger TVD means a larger difference.

### Compute the observed TVD
The next cells show how `.diff()` works and calculate the TVD between the California and UCSD distributions. This is the **observed test statistic** to compare with simulated TVDs.

```python
# The diff method finds the differences of consecutive elements in a Series.
pd.Series([4, 5, -2]).diff()
```

```python
observed_tvd = eth.diff(axis=1).abs().sum().iloc[1] / 2
observed_tvd
```

### Simulation plan
Sample 30,000 students from the California distribution many times, compute each sample’s TVD, and compare the observed TVD with the simulated values.

```python
# Number of students at UCSD in this example.
N_STUDENTS = 30_000
```

```python
eth['California']
```

```python
np.random.multinomial(N_STUDENTS, eth['California'])
```

```python
np.random.multinomial(N_STUDENTS, eth['California']) / N_STUDENTS
```

```python
eth_draws = np.random.multinomial(N_STUDENTS, eth['California'], size=100_000) / N_STUDENTS
eth_draws
```

```python
eth_draws.shape
```

```python
# The values here appear rounded.
tvds = np.abs(eth_draws - eth['California'].to_numpy()).sum(axis=1) / 2
tvds
```

```python
tvds.shape
```

### Visualizing the empirical distribution of the test statistic

### Compare the observed result with simulations
Plot the simulated TVDs, then calculate the fraction that are **at least as large** as the observed TVD. That fraction is the simulated p-value for this test.

```python
observed_tvd
```

```python
fig = px.histogram(pd.DataFrame(tvds), x=0, nbins=20, histnorm='probability', 
                   title='Empirical Distribution of the TVD')
fig
```

```python
(np.array(tvds) >= observed_tvd).mean()
```

### Conclusion
Compare the observed TVD with the simulated null distribution and the resulting p-value. A very small p-value provides evidence against $H_0$.

### Summary of the method

To assess whether an "observed sample" was drawn randomly from a known categorical distribution:
* Use the TVD as the test statistic because it measures the distance between two categorical distributions.
* Sample at random from the population. Compute the TVD between each random sample and the known distribution to get an idea for what reasonable deviations from the eligible pool look like. Repeat this process many, many times.
* Compare:
    - the empirical distribution of TVDs, with
    - the observed TVD from the sample.

```python
eth
```

```python
def ethnicity_test(N_STUDENTS):
    eth_draws = np.random.multinomial(N_STUDENTS, eth['California'], size=100_000) / N_STUDENTS
    tvds = np.sum(np.abs(eth_draws - eth['California'].to_numpy()), axis=1) / 2
    return (np.array(tvds) >= observed_tvd).mean()
```

```python
for i in range(1, 9):
    N_STUDENTS = 2 ** i
    print(f'With {N_STUDENTS} students, the p-value is {ethnicity_test(N_STUDENTS)}.')
```

## 3. Permutation testing

### Hypothesis testing vs. permutation testing
The previous test compares one sample with a known distribution. A **permutation test** compares groups by shuffling their labels under $H_0$ and checking whether the observed difference is unusual.

### Demo 1: Birth weight and smoking
Compare the birth weights of babies whose mothers smoked with those whose mothers did not. Each row represents one mother–baby pair.

```python
baby = pd.read_csv(Path('data') / 'babyweights.csv')
baby
```

```python
baby = baby[['Maternal Smoker', 'Birth Weight']]
baby.head()
```

### Exploratory data analysis

How many babies are in each group? What is the average birth weight within each group?

```python
baby.groupby('Maternal Smoker')['Birth Weight'].agg(['mean', 'count'])
```

### Visualizing birth weight distributions

Below, we draw the distributions of both sets of birth weights.

```python
fig = px.histogram(baby, color='Maternal Smoker', histnorm='probability', marginal='box', 
                   title="Birth Weight by Mother's Smoking Status", barmode='overlay', opacity=0.7)
fig
```

### Hypotheses
- **$H_0$:** Birth-weight distributions are the same in the two groups.
- **$H_1$:** Birth-weight distributions differ between the groups.

```python
fig = px.histogram(baby, color='Maternal Smoker', histnorm='probability', marginal='box', 
                   title="Birth Weight by Mother's Smoking Status", barmode='overlay', opacity=0.7)
fig
```

### Test statistic
Use the difference in mean birth weight: **smokers’ group mean − non-smokers’ group mean**.

```python
group_means = baby.groupby('Maternal Smoker')['Birth Weight'].mean()
group_means
```

```python
group_means.loc[True] - group_means.loc[False]
```

### Permutation-test setup
Under $H_0$, the group labels and birth weights are unrelated. Shuffle weights across the existing labels, recompute the mean difference, and repeat to form a null distribution.

```python
baby.head()
```

### Permutation test
Randomly shuffle the birth weights while keeping the smoker/non-smoker group sizes fixed. Compare the observed mean difference with differences from many shuffles.

### Shuffling

- Our goal, by shuffling, is to randomly assign values in the `'Maternal Smoker'` column to values in the `'Birth Weight'` column.

- We can do this by shuffling either column **independently**.

- Easiest solution: `np.random.permutation`.
    - We could also use `df.sample`, but it's more complicated (and slower).

```python
np.random.permutation(baby['Birth Weight'])
```

```python
with_shuffled = baby.assign(Shuffled_Weights=np.random.permutation(baby['Birth Weight']))
with_shuffled.head()
```

```python
group_means = with_shuffled.groupby('Maternal Smoker').mean()
group_means
```

```python
for x in ['Birth Weight', 'Shuffled_Weights']:
    diff = group_means.loc[True, x] - group_means.loc[False, x]
    fig = px.histogram(
        with_shuffled, x=x, color='Maternal Smoker', histnorm='probability', marginal='box', 
        title=f"Using the {x} column <br>(difference in means = {diff:.2f})",
        barmode='overlay', opacity=0.7)
    fig.update_layout(margin=dict(t=60))
    fig.show()
```

### Repeat the shuffle
One shuffle gives one possible difference under $H_0$. Repeat the process to build the null distribution.

```python
n_repetitions = 500

differences = []
for _ in range(n_repetitions):
    
    # Step 1: Shuffle the weights and store them in a DataFrame.
    with_shuffled = baby.assign(Shuffled_Weights=np.random.permutation(baby['Birth Weight']))

    # Step 2: Compute the test statistic.
    # Remember, False (0) comes before True (1),
    # so this computes True - False.
    group_means = (
        with_shuffled
        .groupby('Maternal Smoker')
        .mean()
        .loc[:, 'Shuffled_Weights']
    )
    difference = group_means.loc[True] - group_means.loc[False]
    
    # Step 4: Store the result
    differences.append(difference)
    
differences[:10]
```

### Calculate the birth-weight p-value
First compute the observed mean difference. The next cells compare it with the shuffled differences: use the **two-sided** p-value for the stated hypothesis that the groups *differ*; the one-sided value answers the narrower question of whether the smoker-group mean is lower.

```python
mean_weights = baby.groupby('Maternal Smoker')['Birth Weight'].mean()
observed_difference = mean_weights[True] - mean_weights[False]
observed_difference
```

```python
# one-sided test: smokers' babies lighter (negative difference expected)
p_value = (differences <= observed_difference).mean()

# if you wanted a two-sided test:
p_value_two_sided = (np.abs(differences) >= abs(observed_difference)).mean()
```

```python
print(p_value_two_sided)
```

```python
fig = px.histogram(
    pd.DataFrame(differences), x=0, nbins=50, histnorm='probability', 
    title='Empirical Distribution of the Mean Differences <br> in Birth Weights (Smoker - Non-Smoker)')
fig.add_vline(x=observed_difference, line_color='red')
fig.update_layout(xaxis_range=[-10, 10], margin=dict(t=60))
```

### Interpretation
Compare the observed difference with the simulated differences and interpret the p-value for the stated alternative.

### Caution
This is observational data: an association between maternal smoking and birth weight does **not**, by itself, establish causation.

Compare employment-status distributions for married and unmarried households using the original study data.

### Cleaning the dataset
Decode the numeric categories for marital status, employment status, and gender.

```python
couples.head()
```

```python
empl = [
    'Working as paid employee',
    'Working, self-employed',
    'Not working - on a temporary layoff from a job',
    'Not working - looking for work',
    'Not working - retired',
    'Not working - disabled',
    'Not working - other'
]
```

Compare the distributions of employment status across married and unmarried households.

### Compare categorical distributions
Count employment-status categories in each group, convert counts to within-group proportions, and visualize the two distributions before calculating TVD.

```python
# Note that this is a shortcut to picking a column for values and using aggfunc='count'.
empl_cnts = couples.pivot_table(index='empl_status', columns='mar_status', aggfunc='size')
empl_cnts
```

```python
empl_cnts.sum()
```

```python
cond_distr = empl_cnts / empl_cnts.sum()
cond_distr
```

```python
cond_distr.plot(kind='barh', title='Distribution of Employment Status, Conditional on Household Type', barmode='group')
```

### Permutation-test hypotheses
- **$H_0$:** Employment-status distributions are the same across marital-status groups.
- **$H_1$:** The distributions differ.
Use TVD as the test statistic.

### TVD
For two categorical distributions, TVD is half the sum of their absolute category-by-category differences.

```python
cond_distr
```

```python
(cond_distr['unmarried'] - cond_distr['married']).abs().sum() / 2
```

```python
def tvd_of_groups(df, groups, cats):
    '''groups: the binary column (e.g. married vs. unmarried).
       cats: the categorical column (e.g. employment status).
    '''
    cnts = df.pivot_table(index=cats, columns=groups, aggfunc='size')
    # Normalize each column.
    distr = cnts / cnts.sum()
    # Compute and return the TVD.
    return (distr['unmarried'] - distr['married']).abs().sum() / 2
```

```python
# Same result as above.
observed_tvd = tvd_of_groups(couples, groups='mar_status', cats='empl_status')
observed_tvd
```

### Simulation
Shuffle marital-status labels, recompute the TVD, and repeat to build a null distribution.

```python
couples.head()
```

```python
couples.assign(shuffled_mar=np.random.permutation(couples['mar_status']))
```

```python
N = 1000
tvds = []

for _ in range(N):
    # Shuffle marital statuses.
    with_shuffled = couples.assign(shuffled_mar=np.random.permutation(couples['mar_status']))
    
    # Compute and store the TVD.
    tvd = tvd_of_groups(with_shuffled, groups='shuffled_mar', cats='empl_status')
    tvds.append(tvd)
```

### Examine the permutation results
The histogram shows TVDs from shuffled marital-status labels. Compare the observed TVD with this distribution to judge whether the group difference is unusual under $H_0$.

```python
fig = px.histogram(tvds, x=0, nbins=50, histnorm='probability', 
                   title='Empirical Distribution of the TVD')
fig.update_layout(xaxis_range=[0, 0.2])
```

### Interpretation
Compare the observed TVD with the shuffled TVDs. A significant association does not explain why the groups differ.

### Summary
Use simulation to evaluate an observed statistic under $H_0$. For two-group comparisons, permutation tests shuffle labels; choose a statistic suited to the data (mean difference for numerical values, TVD for categorical values).
