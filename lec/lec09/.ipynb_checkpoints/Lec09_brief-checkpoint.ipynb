## 1. Categorical hypothesis test: California and UCSD
**Question:** Is the observed UCSD ethnic distribution consistent with random sampling from the California distribution?

- $H_0$: UCSD students follow the California population distribution.
- $H_1$: The distributions differ.
- **Statistic:** Total variation distance (TVD), half the sum of absolute differences between category proportions. A larger TVD indicates a larger difference.

*The proportions in this teaching example come from the original notebook; do not interpret them as current demographic estimates.*

```python
eth = pd.DataFrame(
    [['Asian', 0.15, 0.51],
     ['Black', 0.05, 0.02],
     ['Latino', 0.39, 0.16],
     ['White', 0.35, 0.2],
     ['Other', 0.06, 0.11]],
    columns=['Ethnicity', 'California', 'UCSD']
).set_index('Ethnicity')

eth
```

```python
eth.plot(kind='barh', title='Ethnic Distribution of California and UCSD', barmode='group')
```

```python
observed_tvd = eth.diff(axis=1).abs().sum().iloc[1] / 2
observed_tvd
```

**Simulate under $H_0$:** Draw many samples of 30,000 from the California distribution. Compare each simulated TVD with the observed TVD; the fraction at least as large is the estimated p-value.

```python
# Number of students at UCSD in this example.
N_STUDENTS = 30_000
```

```python
eth_draws = np.random.multinomial(N_STUDENTS, eth['California'], size=100_000) / N_STUDENTS
eth_draws
```

```python
# The values here appear rounded.
tvds = np.abs(eth_draws - eth['California'].to_numpy()).sum(axis=1) / 2
tvds
```

```python
fig = px.histogram(pd.DataFrame(tvds), x=0, nbins=20, histnorm='probability', 
                   title='Empirical Distribution of the TVD')
fig
```

```python
(np.array(tvds) >= observed_tvd).mean()
```

**Interpretation:** If the estimated p-value is at most 0.05, reject $H_0$; otherwise, fail to reject it. A small p-value means the observed TVD would be unusual under the California sampling model, not that the probability of $H_0$ is small.

## 2. Permutation test: Birth weight and maternal smoking
**Question:** Do birth weights differ between babies of mothers who smoke and those who do not?

- $H_0$: Birth-weight distributions are the same in the two groups.
- $H_1$: Babies of mothers who smoke have lower mean birth weight.
- **Statistic:** Mean weight (smoker group) minus mean weight (non-smoker group). More negative values support $H_1$.

Under $H_0$, shuffle the birth weights across the fixed group labels to create a null distribution.

```python
baby = pd.read_csv(Path('data') / 'babyweights.csv')
baby
```

```python
baby = baby[['Maternal Smoker', 'Birth Weight']]
baby.head()
```

```python
baby.groupby('Maternal Smoker')['Birth Weight'].agg(['mean', 'count'])
```

```python
fig = px.histogram(baby, color='Maternal Smoker', histnorm='probability', marginal='box', 
                   title="Birth Weight by Mother's Smoking Status", barmode='overlay', opacity=0.7)
fig
```

```python
group_means = baby.groupby('Maternal Smoker')['Birth Weight'].mean()
group_means
```

```python
group_means.loc[True] - group_means.loc[False]
```

**One shuffle, then many:** Reassign birth weights at random while keeping group sizes fixed. Recalculate the difference in means after every shuffle.

```python
with_shuffled = baby.assign(Shuffled_Weights=np.random.permutation(baby['Birth Weight']))
with_shuffled.head()
```

```python
group_means = with_shuffled.groupby('Maternal Smoker').mean()
group_means
```

```python
n_repetitions = 500

differences = []
for _ in range(n_repetitions):
    
    # Step 1: Shuffle the weights and store them in a DataFrame.
    with_shuffled = baby.assign(Shuffled_Weights=np.random.permutation(baby['Birth Weight']))

    # Step 2: Compute the test statistic.
    # Remember, False (0) comes before True (1),
    # so this computes True - False.
    group_means = (
        with_shuffled
        .groupby('Maternal Smoker')
        .mean()
        .loc[:, 'Shuffled_Weights']
    )
    difference = group_means.loc[True] - group_means.loc[False]
    
    # Step 4: Store the result
    differences.append(difference)
    
differences[:10]
```

```python
mean_weights = baby.groupby('Maternal Smoker')['Birth Weight'].mean()
observed_difference = mean_weights[True] - mean_weights[False]
observed_difference
```

```python
# One-sided p-value: fraction of shuffled differences at most as negative as observed
p_value = (np.asarray(differences) <= observed_difference).mean()
print(f'Estimated p-value: {p_value:.4f}')
print('Decision:', 'Reject H0' if p_value <= 0.05 else 'Fail to reject H0')
```

```python
fig = px.histogram(
    pd.DataFrame(differences), x=0, nbins=50, histnorm='probability', 
    title='Empirical Distribution of the Mean Differences <br> in Birth Weights (Smoker - Non-Smoker)')
fig.add_vline(x=observed_difference, line_color='red')
fig.update_layout(xaxis_range=[-10, 10], margin=dict(t=60))
```

**Interpretation:** Compare the estimated p-value with 0.05. An association in this observational dataset does **not** establish that smoking causes a change in birth weight; other factors may affect both.

## 3. Optional practice: Employment status and household type
**Question:** Do married and unmarried people in this dataset have the same distribution of employment status?

- $H_0$: The employment-status distributions are the same.
- $H_1$: They differ.
- **Statistic:** TVD between the two employment-status distributions. Shuffle marital-status labels and recompute TVD to estimate a p-value.

*Complete this section in class if time permits; otherwise, practice independently.*

```python
couples = pd.read_csv(Path('data') / 'married_couples.csv')
couples.head()
```

```python
couples = couples[['mar_status', 'empl_status', 'gender', 'age']]
couples.head()
```

```python
empl = [
    'Working as paid employee',
    'Working, self-employed',
    'Not working - on a temporary layoff from a job',
    'Not working - looking for work',
    'Not working - retired',
    'Not working - disabled',
    'Not working - other'
]
```

```python
couples = couples.replace({
    'mar_status': {1: 'married', 2: 'unmarried'},
    'gender': {1: 'M', 2: 'F'},
    'empl_status': {(k + 1): empl[k] for k in range(len(empl))}
})
```

```python
# Note that this is a shortcut to picking a column for values and using aggfunc='count'.
empl_cnts = couples.pivot_table(index='empl_status', columns='mar_status', aggfunc='size')
empl_cnts
```

```python
cond_distr = empl_cnts / empl_cnts.sum()
cond_distr
```

```python
def tvd_of_groups(df, groups, cats):
    '''groups: the binary column (e.g. married vs. unmarried).
       cats: the categorical column (e.g. employment status).
    '''
    cnts = df.pivot_table(index=cats, columns=groups, aggfunc='size')
    # Normalize each column.
    distr = cnts / cnts.sum()
    # Compute and return the TVD.
    return (distr['unmarried'] - distr['married']).abs().sum() / 2
```

```python
# Same result as above.
observed_tvd = tvd_of_groups(couples, groups='mar_status', cats='empl_status')
observed_tvd
```

```python
N = 1000
tvds = []

for _ in range(N):
    # Shuffle marital statuses.
    with_shuffled = couples.assign(shuffled_mar=np.random.permutation(couples['mar_status']))
    
    # Compute and store the TVD.
    tvd = tvd_of_groups(with_shuffled, groups='shuffled_mar', cats='empl_status')
    tvds.append(tvd)
```

```python
fig = px.histogram(tvds, x=0, nbins=50, histnorm='probability', 
                   title='Empirical Distribution of the TVD')
fig.update_layout(xaxis_range=[0, 0.2])
```

**Practice decision:** Estimate the p-value as the fraction of permuted TVDs at least as large as the observed TVD. State your decision at $\alpha=0.05$ and explain why this observational comparison does not establish causation.

```python
p_value_couples = (np.asarray(tvds) >= observed_tvd).mean()
print(f'Estimated p-value: {p_value_couples:.4f}')
print('Decision:', 'Reject H0' if p_value_couples <= 0.05 else 'Fail to reject H0')
```
